### Core Matplotlib Concepts

### Basic Data Generation for Practice

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Generate random data for scatter plots
x_data = np.random.random(30) * 100 # Generate 50 random x values between 0 and 100
y_data= np.random.random(30) * 100 # Generate 50 random y values between 0 and 100
```

```python
# Generate time series data for line plots  
years = [2006 + x for x in range(16)]  # 2006-2021
weights = [80, 83, 84, 85, 86, 82, 81, 79, 83, 80, 82, 83, 81, 80, 79,80]
```

```python
# Generate categorical data for bar plots
languages = ['C++', 'C#', 'Python', 'Java', 'Go']
votes = [20, 50, 140, 90, 45]
```

### Plot Types and Applications

#### 1. Scatter Plots

```python
# Basic scatter plot
plt.scatter(x_data, y_data)
plt.show()
```

```python
# Customized scatter plot
plt.scatter(x_data, y_data, 
           color='black',           # Color
           s=50,                # size 
           marker='^',        # Marker style  
           alpha=1)         # Transparency
plt.show()
```

```python
plt.scatter(years, weights)
```

#### 2. Line chart

```python
# Basic line chart
plt.plot(years, weights)
plt.show()
scatter=plt.scatter(years, weights)
plt.show()
```

```python
# Customized line chart
plt.plot(years, weights,
         color='blue',      # or 'b' for short
         linewidth=3,       # or 'lw=3'
         linestyle='--')    # Dashed line
plt.show()
```

```python
# Customized line with scattre plot
plt.plot(years, weights,
         color='black',      # or 'b' for short
         linewidth=2,       # or 'lw=3' thickness
         marker='o',        # Circle markers
         markersize=7,         # Marker size
         markerfacecolor='red',  # Color of the marker face
         linestyle='--')    # Dashed line
plt.show()
```

```python
# Alternative syntax
plt.plot(years, weights, 'y-', lw=2)  # Red dashed line, width 3
plt.show()
```

#### 3. Bar charts

```python
# Basic bar chart
plt.bar(languages, votes)
plt.show()
```

```python
# Customized bar chart
plt.bar(languages, votes,
        color='cyan',         # Bar color
        align='center',       # Bar alignment
        width=0.5,            # Bar width
        edgecolor='black',    # Edge color
        linewidth=1)          # Edge width
plt.show()
```

#### 4. Histogram

```python
# Generate normal distribution data
ages = np.random.normal(20, 1.5, 1000)  # mean=20, std=1.5, n=1000# Basic histogram
plt.hist(ages)
plt.show()
```

```python
# Customized histogram
plt.hist(ages, 
         bins=100,              # Number of bins, bins are the intervals for grouping data 
         color='purple',        # Bar color
         edgecolor='black',    # Edge color
         alpha=0.5,
         cumulative=False)      # Cumulative distribution
plt.show()
```

```python
# Custom bins
custom_bins = [ages.min(), 18, 21, ages.max()] # this will create bins for ages below 18, between 18 and 21, and above 21
plt.hist(ages, bins=custom_bins, edgecolor='black', color='orange', alpha=0.7)
plt.show()
```

#### 5. Pie Chart

```python
# Basic pie chart
plt.pie(votes, labels=languages)
plt.show()
```

```python
# Highly customized pie chart
explode = [0.1, 0, 0, 0, 0]  # Explode first slice
plt.pie(votes, 
        labels=languages,
        explode=explode,
        autopct='%1.2f%%',    # Show percentages
        pctdistance=1.2,      # Distance of percentages
        startangle=90)        # Starting angle
plt.show()
```

#### 6. Box Plots

```python
# Generate sample data
heights = np.random.normal(172, 8, 300)  # mean=172cm, std=8cm # Basic box plot
plt.boxplot(heights)
plt.show()
```

```python
# Multiple box plots
data = [np.random.normal(0, 1, 100),     # Dataset 1 # this line generates a normal distribution with mean=0 and std=1
        np.random.normal(2, 1.5, 100),  # Dataset 2  
        np.random.normal(-1, 0.5, 100)] # Dataset 3
plt.boxplot(data)
plt.show()
```

#### Multiple Figures

```python
# Generate sample data
x1, y1 = np.random.random(100), np.random.random(100) #this line generates 100 random x and y values between 0 and 1
x2 = np.arange(100)    #this line generates an array of integers from 0 to 99
y2 = np.random.random(100)
```

#### Subplots - Multiple Plots in One Figure

```python
import numpy as np
import matplotlib.pyplot as plt

# Create 2x2 subplot grid
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Generate sample data
x = np.arange(100)

# Plot on individual subplots
axes[0,0].plot(x, np.sin(x))
axes[0,0].set_title('Sine Wave')

axes[0,1].plot(x, np.cos(x))  
axes[0,1].set_title('Cosine Wave')

axes[1,0].plot(x, np.random.random(100))
axes[1,0].set_title('Random Function')

axes[1,1].plot(x, np.log(x + 5))
axes[1,1].set_title('Log Function')



# Add overall title
fig.suptitle('Four Mathematical Functions', fontsize=16)

# Adjust layout to prevent overlap (important!)
plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()

```

#### 3D Scatter Plots

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')

# Generate 3D data
x = np.random.random(100) #this line generates 100 random x values between 0 and 1
y = np.random.random(100) 
z = np.random.random(100)

# Create 3D scatter plot
ax.scatter(x, y, z, c=z, cmap='viridis')  # Color by z-value
 # c=z means the color of each point is determined by its z-value 
 # #cmap='viridis' specifies the colormap to use
ax.set_xlabel('X Label')
ax.set_ylabel('Y Label')
ax.set_zlabel('Z Label')
ax.set_title('3D Scatter Plot')
plt.show()
```

#### 3D Line Plots

```python
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')
# Generate 3D curve data
t = np.arange(0, 50, 0.1)
x = np.sin(t)
y = np.cos(t)
z = t

ax.plot(x, y, z, linewidth=2, label='Helix')
ax.set_title('3D Helix')
ax.legend()
plt.show()
```

#### 3D Surface Plots

```python

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(projection='3d')  # Create mesh grid for surface
x = np.arange(-5, 5, 0.1)  #creates an array of x values from -5 to 5 with a step of 0.1
y = np.arange(-5, 5, 0.1)
X, Y = np.meshgrid(x, y)    # Create a mesh grid from x and y arrays
Z = np.sin(X) * np.cos(Y)

# Create surface plot
ax.plot_surface(X, Y, Z, alpha=0.8)
ax.set_title('3D Surface Plot')

# Customize view angle
ax.view_init(azim=45, elev=30)  # Azimuth and elevation
plt.show()
```

### Animations

#### Basic Animation Concept 

```python
import random

# Animated coin flip simulation
heads_tails = [0, 0]  # [heads_count, tails_count]

for i in range(2):
    # Simulate coin flip
    result = random.randint(0, 1)  # 0=heads, 1=tails
    heads_tails[result] += 1
    
    # Clear previous plot and create new one
    plt.clf()  # Clear figure
    plt.bar(['Heads', 'Tails'], heads_tails, color=['red', 'blue'])
    plt.title(f'Coin Flips: {i+1} total')
    plt.ylabel('Count')
    
    # Animate with pause
    plt.pause(1)  # Short pause between frames

plt.show()  # Final static display
print(heads_tails)
```

### Best Practices and Tips

```python
# Good practice: Organize plotting code
def create_scatter_plot(x, y, title="Scatter Plot"):
    """Create a customized scatter plot"""
    plt.figure(figsize=(8, 6))
    plt.scatter(x, y, alpha=0.6, s=50)
    plt.title(title, fontsize=14)
    plt.xlabel('X Values')
    plt.ylabel('Y Values')
    plt.grid(True, alpha=0.3)
    return plt.gcf()  # Return figure object# Usage
fig = create_scatter_plot(x_data, y_data, "My Data Analysis")
fig.savefig('scatter_analysis.png', dpi=300)
plt.show()
```

```python
# Create publication-ready plots
plt.figure(figsize=(10, 6))
plt.plot(x, y, linewidth=2.5, color="#B516D4")
plt.title('Professional Plot Title', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('X-axis Label', fontsize=12)
plt.ylabel('Y-axis Label', fontsize=12)
plt.grid(True, alpha=0.3, linestyle='--')

# Remove top and right spines for cleaner look
ax = plt.gca()          # this line gets the current Axes instance on the current figure
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Adjust margins
plt.tight_layout()  # Automatically adjust subplot params for better layout
plt.savefig('professional_plot.png', dpi=300, bbox_inches='tight')
plt.show()
```

#### Working with Pandas DataFrames

```python
import pandas as pd

# Matplotlib integrates seamlessly with pandas
df = pd.DataFrame({
    'date': pd.date_range('2020-01-01', periods=100),
    'sales': np.random.randn(100).cumsum() + 100,
    'profit': np.random.randn(100).cumsum() + 50
})

# Direct plotting from pandas (uses matplotlib backend)
df.plot(x='date', y=['sales', 'profit'], figsize=(12, 6))
plt.title('Sales and Profit Over Time')
plt.show()
```
