## Filtering Data

### Basic Filtering

```python
import pandas as pd
df = pd.read_csv('./data/bios.csv')
df.head()
df.info()
```

```python
df.loc[df["height_cm"] > 220]

df.loc[df["height_cm"] > 220,["name","height_cm","NOC"]]

height_210 = df.loc[
    (df["height_cm"] > 210) & (df["NOC"] == "United States"),
    ["name", "height_cm"]
]
height_210
```

```python
# Simple conditions
tall_athletes = df[df['height_cm'] > 200]
IND_athletes = df[df['born_country'] == 'IND']
IND_athletes
```

```python
df[df["name"].str.contains("silva", case=False)] # it gives all names with gowda or Gowda
```

### Multiple Conditions

```python
# Use parentheses and & (and) or | (or)
filtered = df[(df['height_cm'] > 200) & (df['born_country'] == 'USA')]
filtered = df[(df['born_city'].str.contains("Bengaluru")) & (df['died_date'] != "Nan")]
filtered
```

```python
res = df[(df["weight_kg"] > 150) & (df["born_country"].str.contains("CH", case=False) | df["born_country"].str.contains("Usa", case=False))]
res
```

### String Operations

```python
# String methods with .str accessor
names_with_keith = df[df['name'].str.contains('Keith', case=False)]
starts_with_a = df[df['name'].str.startswith('A')]
ends_with_son = df[df['name'].str.endswith('son')]
ends_with_son
```

### Advanced String Filtering with Regex

```python
# Regular expressions for complex patterns
vowel_cities = df[df['born_city'].str.contains(r'^[AEIOU]', case=False, na=False)]  # Starts with vowel na= False because some values are NaN. na is used to ignore NaN values
double_letters = df[df['name'].str.contains(r'(.)\1', na=False)]  # Consecutive same letters
hyphenated_names = df[df['name'].str.contains(r'[-\']', na=False)]  # Contains hyphen or apostrophe
vowel_cities
```

## Advanced Functionality

### Shift Operations

```python
# Compare with previous period

 #shift will shift values down by 2 rows
coffee['yesterday_revenue'] = coffee['revenue'].shift(2)
coffee
```

```python
coffee['change'] = coffee['revenue'] - coffee['yesterday_revenue']
coffee
```

```python
coffee['percent_change'] = ((coffee['revenue'] / coffee['yesterday_revenue'] - 1) * 100 ).round(1)
coffee
```

### Ranking

```python
coffee = coffee.drop(coffee.columns[-5:], axis=1)

coffee
```

```python
# Rank values (1 = highest by default)
coffee['most revenue'] = coffee['revenue'].rank(ascending=False)
coffee

```

```python
df.head()
```

```python
avg_height = df.groupby("born_country")["height_cm"].mean()
avg_height
```

```python
ranked = avg_height.rank(ascending=False)  # Higher height = Rank 1
result = pd.DataFrame({
    "avg_height": avg_height,
    "rank": ranked
}).sort_values("rank")

ranked
result
```
